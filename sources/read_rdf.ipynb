```python
# usage: python run.py kg.ttl q.sparql
import sys
from rdflib import Graph
import pandas as pd

ttl = "./../inputs/saref_large.ttl"
q = """
PREFIX ns3: <https://w3id.org/saref#>
PREFIX owl1: <http://www.w3.org/2002/07/ow#/>
PREFIX rdfs1: <http://www.w3.org/2000/01/rdf-schema#>
PREFIX xsd: <http://www.w3.org/2001/XMLSchema#>

SELECT ?state 
WHERE {
    ns3:TempSensInst a ns3:TemperatureSensor ;
                     ns3:hasState ?state .
    ?command a ns3:StartCommand ;
             ns3:actsUpon ?state .
}



"""
g = Graph().parse(ttl, format="turtle")
res = g.query(q)
res

# rdflib SPARQLResult tipini kontrol et
rtype = getattr(res, "type", None)

if rtype == "ASK":
    # ASK sorgularında sonuç boolean’dır
    print("=== ASK RESULT ===")
    print("TRUE" if bool(res) else "FALSE")

elif rtype == "SELECT":
    # SELECT sorgularında vars ve satırlar vardır
    cols = [str(v) for v in res.vars]
    rows = []
    for r in res:
        # r, tuple benzeri bir row; None olabilir -> boş stringe çevir
        rows.append([("" if val is None else str(val)) for val in r])
    df = pd.DataFrame(rows, columns=cols)
    print("=== SELECT RESULT (preview) ===")
    print(df.to_string(index=False, max_cols=200, max_rows=50))
    # İstersen CSV’ye yaz:
    # df.to_csv("out.csv", index=False)
```
